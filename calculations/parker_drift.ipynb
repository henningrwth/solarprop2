**Calculation of the gradient and curvature drift velocities for the `Burger2012` and `Strauss2012` models, which use a simple Parker spiral for the magnetic field.**

```python
import sympy as sp
```

```python
A, B0, r, r0, Be, N = sp.symbols('A B_0 r r_0 B_e N', positive=True)
```

```python
theta, phi = sp.symbols('theta phi', real=True)
```

```python
p, beta, c, q = sp.symbols('p beta c q', real=True)
```

```python
%config InteractiveShell.ast_node_interactivity='last_expr_or_assign'
```

Definition of the magnetic field as a Parker spiral:

$$\mathbf{B}=B_r\mathbf{e}_r + B_\phi\mathbf{e}_\phi =
AB_0\frac{r_0^2}{r^2}\left[\,\mathbf{e}_r - \Gamma\,\mathbf{e}_\phi \right] \times \left[1-2H(\theta-\theta^\prime)\right].$$

```python
Br = sp.Function('B_r', positive=True)(r)
```

```python
Bp = sp.Function('B_phi')(r, theta)
```

Sign function $h=1-2H(\theta-\theta^\prime)$:

```python
h = sp.symbols('h', integer=True)
```

Explicit components of magnetic field:

```python
Brx = A*h*B0*r0**2/r**2
```

```python
G = sp.Function('Gamma')(r, theta)
```

```python
Bpx = -Brx*G
```

```python
Bsqr = Br**2+Bp**2
```

```python
components = [(Br, Brx), (Bp, Bpx)]
```

$$\langle \vec{v}_{dr}\rangle = \frac{p\beta c}{3q}\mathrm{rot}\frac{\vec{B}}{B^2}$$

$$\vec{F}=F_r\vec{e}_r+F_\theta\vec{e}_\theta+F_\phi\vec{e}_\phi$$ 

$$\mathrm{rot}\vec{F}=\frac{1}{r\sin\theta}\left[\frac{\partial}{\partial\theta}(F_\phi\sin\theta)-\frac{\partial F_\theta}{\partial\phi}\right]\vec{e}_r + \left[\frac{1}{r\sin\theta}\frac{\partial F_r}{\partial\phi}-\frac{1}{r}\frac{\partial}{\partial r}(rF_\phi)\right]\vec{e}_\theta+
\left[\frac{1}{r}\frac{\partial}{\partial r}(rF_\theta)-\frac{1}{r}\frac{\partial F_r}{\partial\theta}\right]\vec{e}_\phi$$

```python
F_r = Br/Bsqr
```

```python
F_r = sp.simplify(F_r.subs(components))
```

```python
F_phi = sp.simplify((Bp/Bsqr).subs(components))
```

```python
f = p*beta*c/3/q
```

```python
Omega = sp.symbols('Omega', real=True)
```

```python
v = sp.Function('v', positive=True)(theta)
```

```python
Gf = Omega/v*r*sp.sin(theta)
```

$$\frac{\partial\Gamma}{\partial r}=\frac{\Gamma}{r}$$

```python
sp.diff(Gf, r)
```

```python
sp.diff(Gf, r) - Gf/r
```

$$\frac{\partial\Gamma}{\partial\theta}=\Gamma\cot\theta-\Gamma v^\prime/v$$

```python
sp.diff(Gf, theta)
```

```python
sp.simplify(sp.diff(Gf, theta)- (Gf*sp.cot(theta)-Gf*sp.diff(v, theta)/v))
```

```python
vprime = sp.symbols('vprime', real=True)
```

```python
sp.Eq(N, 1+G**2)
```

```python
rotF_r = sp.simplify(1/r/sp.sin(theta)*(sp.diff(F_phi*sp.sin(theta), theta)))
```

```python
rotF_r_2 = sp.simplify(rotF_r.subs(G**2+1, N))
```

```python
rotF_r_3 = sp.simplify(rotF_r_2.subs([(sp.Derivative(G, theta), G*sp.cot(theta) - G*vprime/v)]))
```

```python
v_dr_r = f*rotF_r_3
```

```python
v_dr_r_0 = v_dr_r.subs(vprime, 0)
```

```python
rotF_theta = sp.simplify(1/r/sp.sin(theta)*sp.diff(F_r, phi) - 1/r*sp.diff(r*F_phi, r))
```

```python
rotF_theta_2 = sp.simplify(rotF_theta.subs(sp.Derivative(G, r), G/r))
```

```python
rotF_theta_3 = sp.simplify(rotF_theta_2.subs(G**2+1, N))
```

```python
v_dr_theta = f*rotF_theta_3
```

```python
rotF_phi = sp.simplify(-1/r*sp.diff(F_r, theta))
```

```python
rotF_phi_2 = rotF_phi.subs([(sp.Derivative(G, theta), G*sp.cot(theta) - G*vprime/v), (G**2+1, N)])
```

```python
v_dr_phi = f*rotF_phi_2
```

```python
v_dr_phi_0 = v_dr_phi.subs(vprime, 0)
```
