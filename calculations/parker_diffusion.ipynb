**Calculation of diffusion coefficients and their relevant derivatives for the `Burger2012` and `Strauss2012` models, which use a simple Parker spiral for the magnetic field.**

```python
import sympy as sp
```

```python
x, f, r, theta = sp.symbols('x f r theta', positive=True)
```

```python
A, B0, r0, Be = sp.symbols(r'A B_0 r_0 B_\oplus', positive=True)
```

```python
mu, phi = sp.symbols('mu phi', real=True)
```

```python
%config InteractiveShell.ast_node_interactivity='last_expr_or_assign'
```

Let $x=\tan(w)$. Then $\cos^2(w)=\cos^2(\arctan(x))=\frac{1}{x^2+1}$, etc.

```python
c2 = sp.simplify(sp.cos(sp.atan(x))**2)
```

```python
c1 = sp.simplify(sp.cos(sp.atan(x)))
```

```python
s2 = sp.simplify(sp.sin(sp.atan(x))**2)
```

```python
s1 = sp.simplify(sp.sin(sp.atan(x)))
```

Definition of the magnetic field as a Parker spiral:

$$\mathbf{B}=B_r\mathbf{e}_r + B_\phi\mathbf{e}_\phi =
AB_0\frac{r_0^2}{r^2}\left[\,\mathbf{e}_r - \Gamma\,\mathbf{e}_\phi \right] \times \left[1-2H(\theta-\theta^\prime)\right].$$

```python
Br = sp.Function('B_r', positive=True)(r)
```

We use $\mu=\cos\theta$:

```python
Bp = sp.Function('B_phi')(r, mu)
```

Calculation of $\tan\psi=-B_\phi/B_r$:

```python
tanpsi = -Bp/Br
```

Calculation of $\cos^2\psi$:

```python
cos2psi = sp.simplify(c2.subs(x, tanpsi))
```

Calculation of $\sin^2\psi$:

```python
sin2psi = sp.simplify(s2.subs(x, tanpsi))
```

Sign function $h=1-2H(\theta-\theta^\prime)$:

```python
h = sp.symbols('h', integer=True)
```

Explicit components of magnetic field:

```python
Brx = A*h*B0*r0**2/r**2
```

```python
G = sp.Function('Gamma', positive=True)(r, mu)
```

```python
Bpx = -Brx*G
```

```python
components = [(Br, Brx), (Bp, Bpx)]
```

```python
sin2psir = sp.simplify(sin2psi.subs(components))
```

```python
cos2psir = sp.simplify(cos2psi.subs(components))
```

```python
B = sp.sqrt(Br**2+Bp**2)
```

```python
BeoverB = Be/sp.simplify(B.subs(components)).subs(sp.Abs(h), 1)
```

```python
kappa_par, kappa_perp, kappa_perp_r, kappa_perp_theta = sp.symbols(r'kappa_\parallel kappa_\perp kappa_\perp_r kappa_\perp_theta', positive=True)
```

$\kappa_{rr}=\kappa_\parallel\cos^2\psi+\kappa_{\perp r}\sin^2\psi$:

```python
kappa_rr = sp.simplify(kappa_par*cos2psir+kappa_perp_r*sin2psir)
```

Explicit expression, assuming $\kappa_{\perp r}=\kappa_{\perp\theta}\equiv\rho\kappa_\parallel$:

```python
rho = sp.symbols('rho', positive=True)
```

Factor in front of diffusion coefficient: $f=\lambda_0v K_P(P)/3$, or $f=\beta\kappa_0K_P(P)$, respectively, so that $\kappa_\parallel=f\frac{B_\oplus}{B}$.

```python
Krr = sp.simplify(kappa_rr.subs(kappa_perp_r, rho*kappa_par).subs(kappa_par, f*BeoverB))
```

$\kappa_{r\phi}=\kappa_{\phi r}=(\kappa_{\perp r}-\kappa_\parallel)\cos\psi\sin\psi$

```python
kappa_r_phi = sp.simplify((kappa_perp_r-kappa_par)*sp.sqrt(cos2psir*sin2psir))
```

```python
Krp = sp.simplify(kappa_r_phi.subs(kappa_perp_r, rho*kappa_par).subs(kappa_par, f*BeoverB))
```

$\kappa_{\theta\theta}=\kappa_{\perp\theta}$

```python
kappa_theta_theta = kappa_perp_theta
```

```python
Ktt = sp.simplify(kappa_theta_theta.subs(kappa_perp_theta, rho*kappa_par).subs(kappa_par, f*BeoverB))
```

$\kappa_{\phi\phi}=\kappa_\parallel\sin^2\psi+\kappa_{\perp r}\cos^2\psi$

```python
kappa_phi_phi = sp.simplify(kappa_par*sin2psir + kappa_perp_r*cos2psir)
```

```python
Kpp = sp.simplify(kappa_phi_phi.subs(kappa_perp_r, rho*kappa_par).subs(kappa_par, f*BeoverB))
```

Calculate $\frac{1}{r^2}\frac{\partial}{\partial r}(r^2\kappa_{rr})$:

```python
dr1 = (1/r**2)*sp.simplify(sp.diff(r**2*Krr, r))
```

Since $\Gamma=\frac{\Omega}{v_{sw}}r\sin\theta$, we have $\frac{d}{dr}\Gamma=\Gamma/r$. Simplification with $N(r)=1+\Gamma^2$.

```python
N = sp.symbols('N', positive=True)
```

```python
dr1final = sp.simplify(dr1.subs(sp.Derivative(G, r), G/r)).subs(G**2+1, N)
```

Calculate $\frac{1}{r\sin\theta}\frac{\partial}{\partial\phi}\kappa_{r\phi}$:

```python
dr2 = (1/r/sp.sqrt(1-mu**2))*sp.simplify(sp.diff(Krp, phi))
```

Calculate $\frac{1}{r^2\sin^2\theta}\frac{\partial}{\partial\phi}\kappa_{\phi\phi}$:

```python
dphi2 = (1/r/(1-mu**2))*sp.simplify(sp.diff(Kpp, phi))
```

Calculate $\frac{1}{r^2\sin\theta}\frac{\partial}{\partial r}(r\kappa_{r\phi})$:

```python
dphi1 = (1/r**2/sp.sqrt(1-mu**2))*sp.simplify(sp.diff(r*Krp, r))
```

```python
dphi1final = sp.simplify(dphi1.subs(sp.Derivative(G, r), G/r)).subs(G**2+1, N)
```

Calculate $\frac{\partial}{\partial\mu}\left(\frac{K_{\theta\theta}(1-\mu^2)}{r^2}\right)$:

```python
dmu = (1/r**2) * sp.simplify(sp.diff(Ktt*(1-mu**2), mu))
```

Investigate polar derivatives:

```python
Omega = sp.symbols('Omega', real=True)
```

```python
vtheta = sp.Function('v', positive=True)(theta)
v = sp.Function('v', positive=True)(mu)
```

```python
Gf = Omega/vtheta*r*sp.sin(theta)
```

```python
Gfmu = Gf.subs(sp.sin(theta), sp.sqrt(1-mu**2)).subs(vtheta, v)
```

```python
sp.simplify(-1/sp.sin(theta)*sp.diff(Gf, theta))
```

```python
Gammaprime = sp.simplify(sp.diff(Gfmu, mu))
```

So, $\Gamma^\prime = \frac{\mathrm{d}v}{\mathrm{d}\mu} = -\Gamma\left(\frac{\cos\theta}{\sin^2\theta}+\frac{v^\prime}{v}\right) = -\Gamma\left(\frac{\mu}{1-\mu^2}+\frac{v^\prime}{v}\right)$ with $v^\prime=\frac{\mathrm{d}v(\mu)}{\mathrm{d}\mu}$:

```python
testGammaprime = -Gfmu*(mu/(1-mu**2) + sp.diff(v, mu)/v)
```

```python
sp.simplify(testGammaprime-Gammaprime)
```

```python
Gprime = sp.symbols('Gammaprime', real=True)
```

For the case $v^\prime=0$:

```python
dmu_vr = sp.simplify(dmu.subs(sp.Derivative(G, mu), -G*(mu/(1-mu**2))))
```

```python
dmufinal = dmu_vr.subs(G**2+1, N)
```

Convert to C code:

```python
import sympy.abc as abc
```

```python
codesubs = [(G, abc.gamma), (Be, sp.Symbol('B_e'))]
```

$\frac{1}{r^2}\frac{\partial}{\partial r}(r^2K_{rr})$:

```python
dr1code = sp.simplify(dr1final.subs(codesubs))
```

```python
print(sp.ccode(dr1code))
```

$\frac{\partial}{\partial\mu}\left(\frac{K_{\theta\theta}(1-\mu^2)}{r^2}\right)$:

```python
dmucode = sp.simplify(dmufinal.subs(codesubs))
```

```python
print(sp.ccode(dmucode))
```

$\frac{1}{r^2\sin\theta}\frac{\partial}{\partial r}(r\kappa_{r\phi})$:

```python
dphi1code = sp.simplify(dphi1final.subs(codesubs))
```

```python
print(sp.ccode(dphi1code))
```
