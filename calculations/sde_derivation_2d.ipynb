**Full derivation of the 2D stochastic differential equations used in `SOLARPROP`.**

These equations neglect the azimuthal coordinate, but retain a diffusion tensor component $\kappa_{r\theta}\neq{}0$.

```python
import sympy as sp
```

```python
r, mu, theta = sp.symbols('r mu theta', positive=True)
```

```python
vr, vt, vsw = sp.symbols('V_r V_theta V_sw', positive=True)
```

```python
%config InteractiveShell.ast_node_interactivity='last_expr_or_assign'
```

```python
krr = sp.Function('kappa_rr', positive=True)(r, mu)
```

```python
krt = sp.Function('kappa_r_theta', positive=True)(r, mu)
```

```python
ktt = sp.Function('kappa_theta_theta', positive=True)(r, mu)
```

```python
f = sp.Function('f')(r, theta)
```

We start from Parker's equation:

$$
\frac{\partial f}{\partial t}=\vec{\nabla}\cdot\left(\kappa\cdot\vec{\nabla}f-\vec{V}f\right)+\frac{1}{3p^2}(\vec{\nabla}\cdot\vec{V}_{sw})\frac{\partial p^3f}{\partial p}
$$

where $\vec{V}=\vec{V}_{sw}+\vec{V}_{dr}$.

We write the individual terms in 2D spherical coordinates, and make use of the definitions $\vec{V}_{sw}=V_{sw}\,\vec{e}_r$, $\vec{V}_{dr}\equiv V_{dr}\,\vec{e}_r + V_{d\theta}\,\vec{e}_\theta$ , and $\vec{\nabla}\cdot\vec{V}_{dr}=0$:

$
\begin{align}
\vec{\nabla}\cdot\left(\kappa\cdot\vec{\nabla}f\right)=\frac{1}{r^2}\frac{\partial}{\partial r}\left(r^2\kappa_{rr}\frac{\partial f}{\partial r}+r\kappa_{r\theta}\frac{\partial f}{\partial\theta}\right)+\frac{1}{r\sin\theta}\frac{\partial}{\partial\theta}\left(\sin\theta\kappa_{\theta r}\frac{\partial f}{\partial r}+\sin\theta\frac{\kappa_{\theta\theta}}{r}\frac{\partial f}{\partial\theta}\right),
\end{align}
$

$
\begin{align}
-\vec{\nabla}\cdot\vec{V}f &= -\vec{\nabla}\cdot\vec{V}_{sw}f-\vec{\nabla}\cdot\vec{V}_{dr}f \\
& = -\frac{1}{r^2}\frac{\partial r^2 V_{sw}f}{\partial r}-f( \vec{\nabla}\cdot\vec{V}_{dr} ) - \vec{V}_{dr}\cdot\vec{\nabla}f \\
& =  -\frac{1}{r^2}\frac{\partial(r^2 V_{sw})}{\partial r}f-V_{sw}\frac{\partial f}{\partial r}-V_{dr}\frac{\partial f}{\partial r}-\frac{V_{d\theta}}{r}\frac{\partial f}{\partial\theta},
\end{align}
$

$
\begin{align}
\frac{1}{3p^2}(\vec{\nabla}\cdot\vec{V}_{sw})\frac{\partial p^3f}{\partial p}&=\frac{1}{r^2}\frac{\partial (r^2V_{sw})}{\partial r}\frac{\partial p^3 f}{\partial p^3}\\
&=\frac{1}{r^2}\frac{\partial (r^2V_{sw})}{\partial r}f + \frac{1}{r^2}\frac{\partial (r^2V_{sw})}{\partial r}p^3\frac{\partial f}{\partial p^3}\\
&=\frac{1}{r^2}\frac{\partial (r^2V_{sw})}{\partial r}f + \frac{p}{3r^2}\frac{\partial (r^2V_{sw})}{\partial r}\frac{\partial f}{\partial p}.
\end{align}
$

We define 

$$
\mu=\cos\theta,
$$

which leads to

$$
\frac{1}{\sin\theta}\frac{\partial}{\partial\theta}=-\frac{\partial}{\partial\mu}
$$

and thus the first term becomes

$
\begin{align}
\vec{\nabla}\cdot\left(\kappa\cdot\vec{\nabla}f\right)&=\frac{1}{r^2}\frac{\partial}{\partial r}\left(r^2\kappa_{rr}\frac{\partial f}{\partial r}-r\kappa_{r\theta}\sqrt{1-\mu^2}\frac{\partial f}{\partial\mu}\right)+\frac{1}{r}\frac{\partial}{\partial\mu}\left(-\sqrt{1-\mu^2}\kappa_{\theta r}\frac{\partial f}{\partial r}+(1-\mu^2)\frac{\kappa_{\theta\theta}}{r}\frac{\partial f}{\partial\mu}\right)\\
&= \kappa_{rr}\frac{\partial^2 f}{\partial r^2} + \frac{1}{r^2} \frac{\partial (r^2\kappa_{rr})}{\partial r}\frac{\partial f}{\partial r} - \frac{\kappa_{r\theta}\sqrt{1-\mu^2}}{r}\frac{\partial^2 f}{\partial r\partial\mu} - \frac{\sqrt{1-\mu^2}}{r^2}\frac{\partial(r\kappa_{r\theta})}{\partial r}\frac{\partial f}{\partial\mu} - \frac{\sqrt{1-\mu^2}\kappa_{\theta r}}{r}\frac{\partial^2 f}{\partial r\partial\mu}-\frac{1}{r}\frac{\partial(\sqrt{1-\mu^2}\kappa_{\theta r})}{\partial\mu}\frac{\partial f}{\partial r}+\frac{(1-\mu^2)\kappa_{\theta\theta}}{r^2}\frac{\partial^2 f}{\partial\mu^2}+\frac{1}{r^2}\frac{\partial((1-\mu^2)\kappa_{\theta\theta})}{\partial\mu}\frac{\partial f}{\partial\mu}.\\
\end{align}
$


The complete transport equation becomes

$$
\frac{\partial f}{\partial t}=\kappa_{rr}\frac{\partial^2 f}{\partial r^2} + \frac{1}{r^2} \frac{\partial (r^2\kappa_{rr})}{\partial r}\frac{\partial f}{\partial r} - \frac{\kappa_{r\theta}\sqrt{1-\mu^2}}{r}\frac{\partial^2 f}{\partial r\partial\mu} - \frac{\sqrt{1-\mu^2}}{r^2}\frac{\partial(r\kappa_{r\theta})}{\partial r}\frac{\partial f}{\partial\mu} - \frac{\sqrt{1-\mu^2}\kappa_{\theta r}}{r}\frac{\partial^2 f}{\partial r\partial\mu}-\frac{1}{r}\frac{\partial(\sqrt{1-\mu^2}\kappa_{\theta r})}{\partial\mu}\frac{\partial f}{\partial r}+\frac{(1-\mu^2)\kappa_{\theta\theta}}{r^2}\frac{\partial^2 f}{\partial\mu^2}+\frac{1}{r^2}\frac{\partial((1-\mu^2)\kappa_{\theta\theta})}{\partial\mu}\frac{\partial f}{\partial\mu}-V_{sw}\frac{\partial f}{\partial r}-V_{dr}\frac{\partial f}{\partial r}+\frac{\sqrt{1-\mu^2}\,V_{d\theta}}{r}\frac{\partial f}{\partial\mu}+ \frac{p}{3r^2}\frac{\partial (r^2V_{sw})}{\partial r}\frac{\partial f}{\partial p}
$$

Therefore, we have expressed the transport equation in the standard form of the backward-in-time Kolmogorov equation:

$$
\frac{\partial f}{\partial t}=\sum\limits_iA_i(\vec{q},t)\frac{\partial f}{\partial q_i}+\frac{1}{2}\sum\limits_{i,j}\left[B(\vec{q},t)B^T(\vec{q},t)\right]_{i,j}\frac{\partial^2 f}{\partial q_i\partial q_j}
$$

with the corresponding SDEs given by

$$
\mathrm{d}\vec{q}=A(\vec{q},t)\mathrm{d}t+B(\vec{q},t)\mathrm{d}W,
$$
where $\mathrm{d}W$ is the increment of the Wiener process.

We use

$$
\vec{q}=(r, \mu, p)
$$

and from comparing the coefficients, we get

$$
A_r = \frac{1}{r^2}\frac{\partial}{\partial r}(r^2\kappa_{rr}) - \frac{1}{r}\frac{\partial}{\partial\mu}(\kappa_{r\theta}\sqrt{1-\mu^2}) - V_{sw} - V_{dr}
$$

$$
A_\mu = -\frac{\sqrt{1-\mu^2}}{r^2}\frac{\partial}{\partial r}(r\kappa_{r\theta}) + \frac{1}{r^2}\frac{\partial}{\partial\mu}(\kappa_{\theta\theta}(1-\mu^2))+\frac{\sqrt{1-\mu^2}}{r}V_{d\theta}
$$

$$
A_p=\frac{p}{3r^2}\frac{\partial r^2V_{sw}}{\partial r}
$$

and

$$
BB^T=
\begin{bmatrix}
2\kappa_{rr} & -\frac{2\kappa_{r\theta}\sqrt{1-\mu^2}}{r} \\
-\frac{2\kappa_{r\theta}\sqrt{1-\mu^2}}{r}& \frac{2\kappa_{\theta\theta}(1-\mu^2)}{r^2}\\
\end{bmatrix}
$$

The elements of the decomposition matrix $B$ are found to be:

```python
B11 = sp.sqrt((krr*ktt-krt**2)/sp.Rational(1,2)/ktt)
```

```python
B12 = -krt*sp.sqrt(2/ktt)
```

```python
B22 = sp.sqrt(2*ktt*(1-mu**2)/r**2)
```

```python
B = sp.Matrix([[B11, B12], [0, B22]])
```

Proof:

```python
sp.simplify(B @ B.T)
```

Dropping the explicit dependence of the diffusion coefficients on $r$ and $\mu$ in the notation, the SDEs are found to be

$$
\mathrm{d}r=\left[\frac{1}{r^2}\frac{\partial}{\partial r}(r^2\kappa_{rr}) - \frac{1}{r}\frac{\partial}{\partial\mu}(\kappa_{r\theta}\sqrt{1-\mu^2}) - V_{sw}-V_{dr}\right]\,\mathrm{d}t + w_r\,\frac{\sqrt{- 2 \kappa_{r \theta}^{2} + 2 \kappa_{rr}\kappa_{\theta \theta}}}{\sqrt{\kappa_{\theta \theta}}}\,\sqrt{\mathrm{d}t} - w_\mu \frac{\sqrt{2} \kappa_{r \theta}}{\sqrt{\kappa_{\theta \theta}}} \,\sqrt{\mathrm{d}t}
$$

$$
\mathrm{d}\mu = \left[ -\frac{\sqrt{1-\mu^2}}{r^2}\frac{\partial}{\partial r}(r\kappa_{r\theta}) + \frac{1}{r^2}\frac{\partial}{\partial\mu}(\kappa_{\theta\theta}(1-\mu^2))+\frac{\sqrt{1-\mu^2}}{r}V_{d\theta} \right]\,\mathrm{d}t + w_\mu \frac{\sqrt{2(1 - \mu^{2})\kappa_{\theta \theta}}}{r} \,\sqrt{\mathrm{d}t}
$$

$$
\mathrm{d}p=\frac{p}{3r^2}\frac{\partial r^2V_{sw}}{\partial r}\,\mathrm{d}t
$$

where $w_r$ and $w_\mu$ are Gaussian random numbers with a mean of zero and unit variance.
