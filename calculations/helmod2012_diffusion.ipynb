**Derivation of the diffusion coefficients in heliospheric polar coordinates and their relevant derivatives, for the modified magnetic field geometry used in the `helmod2012` model.**

```python
import sympy as sp
```

```python
x, f, r, theta, iota, rho = sp.symbols('x f r theta iota rho', positive=True)
```

```python
A, B0, r0, Be = sp.symbols('A B_0 r_0 B_e', positive=True)
```

```python
mu = sp.symbols('mu', real=True)
```

```python
%config InteractiveShell.ast_node_interactivity='last_expr_or_assign'
```

Let $x=\tan(w)$. Then $\cos^2(w)=\cos^2(\arctan(x))=\frac{1}{x^2+1}$, etc.

```python
c2 = sp.simplify(sp.cos(sp.atan(x))**2)
```

```python
c1 = sp.simplify(sp.cos(sp.atan(x)))
```

```python
s2 = sp.simplify(sp.sin(sp.atan(x))**2)
```

```python
s1 = sp.simplify(sp.sin(sp.atan(x)))
```

Definition of the magnetic field after polar extension (Bobik+, Adv. Astron. (2013) 793072):

$$\mathbf{B}=B_r\mathbf{e}_r + B_\phi\mathbf{e}_\phi + B_\theta\mathbf{e}_\theta=
AB_0\frac{r_0^2}{r^2}\left[\,\mathbf{e}_r - \Gamma\,\mathbf{e}_\phi + \frac{\delta(\theta)}{r_\odot}r\,\mathbf{e}_\theta\right] \times \left[1-2H(\theta-\theta^\prime)\right].$$

```python
Br = sp.Function('B_r', positive=True)(r)
```

We use $\mu=\cos\theta$:

```python
Bt = sp.Function('B_theta')(r, mu)
```

```python
Bp = sp.Function('B_phi')(r, mu)
```

Calculation of $\tan\xi=B_\theta/B_r$:

```python
tanxi = Bt/Br
```

Calculation of $\tan\psi=-B_\phi/\sqrt{B_r^2+B_\theta^2}$:

```python
tanpsi = -Bp/sp.sqrt(Br**2+Bt**2)
```

Calculation of $\cos^2\xi$:

```python
cos2xi = sp.simplify(c2.subs(x, tanxi))
```

Calculation of $\sin^2\xi$:

```python
sin2xi = sp.simplify(s2.subs(x, tanxi))
```

Calculation of $\cos\xi$:

```python
cosxi = sp.simplify(sp.sqrt(cos2xi))
```

Calculation of $\sin\xi$:

```python
sinxi = sp.simplify(sp.sqrt(sin2xi))
```

Calculation of $\cos^2\psi$:

```python
cos2psi = sp.simplify(c2.subs(x, tanpsi))
```

Calculation of $\sin^2\psi$:

```python
sin2psi = sp.simplify(s2.subs(x, tanpsi))
```

Sign function $h=1-2H(\theta-\theta^\prime)$:

```python
h = sp.symbols('h', integer=True)
```

Explicit components of magnetic field:

```python
Brx = A*h*B0*r0**2/r**2
```

```python
G = sp.Function('Gamma')(r, mu)
```

```python
Bpx = -Brx*G
```

Define extension function $t(r, \mu)=\frac{\delta(\mu)}{r_\odot}r$.

```python
t = sp.Function('t', positive=True)(r, mu)
```

```python
Btx = Brx*t
```

```python
components = [(Br, Brx), (Bp, Bpx), (Bt, Btx)]
```

```python
sin2psir = sp.simplify(sin2psi.subs(components))
```

```python
cos2psir = sp.simplify(cos2psi.subs(components))
```

```python
sin2xir = sp.simplify(sin2xi.subs(components))
```

```python
cos2xir = sp.simplify(cos2xi.subs(components))
```

```python
cosxir = sp.simplify(cosxi.subs(components))
```

```python
sinxir = sp.simplify(sinxi.subs(components))
```

Factor in front of diffusion coefficient: $f=\beta K_0K_P(P,t)/3$.

```python
B = sp.sqrt(Br**2+Bp**2+Bt**2)
```

```python
BeoverB = Be/sp.simplify(B.subs(components))
```

$K_{rr}=f\frac{B_\oplus}{B}\left(\iota\rho\sin^2\xi+\cos^2\xi\cdot(\cos^2\psi+\rho\sin^2\psi)\right)$:

```python
Krr = sp.simplify(f*BeoverB*(iota*rho*sin2xir+cos2xir*(cos2psir+rho*sin2psir)))
```

Calculate $\frac{1}{r^2}\frac{\partial}{\partial r}(r^2K_{rr})$:

```python
dr1 = (1/r**2)*sp.simplify(sp.diff(r**2*Krr, r))
```

```python
N, tau = sp.symbols('N tau', positive=True)
```

Simplification with $N(r)=1+\Gamma^2+t^2$ and $\tau=1+t^2$:

```python
Krrfinal = Krr.subs(G**2+t**2+1, N).subs(t**2+1, tau).subs(sp.Abs(h), 1)
```

```python
dr1simple = dr1.subs(G**2+t**2+1, N).subs(t**2+1, tau).subs(sp.Abs(h), 1)
```

Since $\Gamma=\frac{\Omega}{v_{sw}}r\sin\theta$ and $t(r)=\frac{\delta(\theta)}{r_\odot}r$, we have $\frac{d}{dr}\Gamma=\Gamma/r$ and $\frac{d}{dr}t(r)=t/r$.

```python
dr1final = sp.simplify(dr1simple.subs([(sp.Derivative(t, r), t/r), (sp.Derivative(G, r), G/r)]))
```

$K_{r\theta}=K^S_{\theta r}=f\frac{B_\oplus}{B}\left(\sin\xi\cos\xi\cdot(\cos^2\psi+\rho\sin^2\psi-\iota\rho)\right)$:

```python
Krtheta = sp.simplify(f*BeoverB*(sinxir*cosxir*(cos2psir+rho*sin2psir-iota*rho)))
```

```python
Krthetafinal = Krtheta.subs(G**2+t**2+1, N).subs(t**2+1, tau)
```

Calculate $-\frac{1}{r^2}\frac{\partial}{\partial r}(rK_{r\theta}\sqrt{1-\mu^2})$:

```python
dmu1 = (-1/r**2)*sp.simplify(sp.diff(r*sp.sqrt(1-mu**2)*Krtheta, r))
```

```python
dmu1simple = dmu1.subs(G**2+t**2+1, N).subs(t**2+1, tau)
```

```python
dmu1final = sp.simplify(dmu1simple.subs([(sp.Derivative(t, r), t/r), (sp.Derivative(G, r), G/r)]))
```

$K_{\theta\theta}=f\frac{B_\oplus}{B}\left(\iota\rho\cos^2\xi+\sin^2\xi(\cos^2\psi+\rho\sin^2\psi)\right)$:

```python
Ktt = sp.simplify(f*BeoverB*(iota*rho*cos2xir + sin2xir*(cos2psir+rho*sin2psir))).subs(sp.Abs(h), 1)
```

```python
Kttfinal = Ktt.subs(G**2+t**2+1, N).subs(t**2+1, tau)
```

Calculate $\frac{\partial}{\partial\mu}\left(\frac{K_{\theta\theta}(1-\mu^2)}{r^2}\right)$:

```python
dmu2 = (1/r**2) * sp.simplify(sp.diff(Ktt*(1-mu**2), mu))
```

```python
dmu2simple = dmu2.subs(G**2+t**2+1, N).subs(t**2+1, tau)
```

Investigate polar derivatives:

```python
Omega = sp.symbols('Omega', real=True)
```

```python
vtheta = sp.Function('v', positive=True)(theta)
v = sp.Function('v', positive=True)(mu)
```

```python
Gf = Omega/vtheta*r*sp.sin(theta)
```

```python
Gfmu = Gf.subs(sp.sin(theta), sp.sqrt(1-mu**2)).subs(vtheta, v)
```

```python
sp.simplify(-1/sp.sin(theta)*sp.diff(Gf, theta))
```

```python
Gammaprime = sp.simplify(sp.diff(Gfmu, mu))
```

So, $\Gamma^\prime = \frac{\mathrm{d}v}{\mathrm{d}\mu} = -\Gamma\left(\frac{\cos\theta}{\sin^2\theta}+\frac{v^\prime}{v}\right) = -\Gamma\left(\frac{\mu}{1-\mu^2}+\frac{v^\prime}{v}\right)$ with $v^\prime=\frac{\mathrm{d}v(\mu)}{\mathrm{d}\mu}$:

```python
testGammaprime = -Gfmu*(mu/(1-mu**2) + sp.diff(v, mu)/v)
```

```python
sp.simplify(testGammaprime-Gammaprime)
```

```python
Gprime = sp.symbols('Gammaprime', real=True)
```

Derivative of polar correction function has to be treated with care, since correction is cut off near the poles.

```python
rS, delta0 = sp.symbols('r_S delta_0', positive=True)
```

```python
delta = sp.Function('delta')(r, mu)
```

```python
tf = delta/rS*r
```

```python
sp.diff(tf, mu)
```

```python
deltaf = h * delta0 / sp.sin(theta)
```

```python
deltafmu = deltaf.subs(sp.sin(theta), sp.sqrt(1-mu**2))
```

```python
sp.diff(deltafmu, mu)
```

```python
tprime = sp.symbols('tprime', real=True)
```

```python
dmu2final = sp.simplify(dmu2simple.subs([(sp.Derivative(G, mu), Gprime), (sp.Derivative(t, mu), tprime)]))
```

```python
dmu2finalt = sp.simplify(dmu2final.subs(1-mu**2, sp.sin(theta)**2))
```

Calculate $-\frac{1}{r}\frac{\partial}{\partial\mu}\left(K_{r\theta}\sqrt{1-\mu^2}\right)$:

```python
dr2 = (-1/r) * sp.simplify(sp.diff(Krtheta*sp.sqrt(1-mu**2), mu))
```

```python
dr2simple = dr2.subs(G**2+t**2+1, N).subs(t**2+1, tau)
```

```python
dr2final = sp.simplify(dr2simple.subs([(sp.Derivative(G, mu), Gprime), (sp.Derivative(t, mu), tprime)]))
```

```python
dr2finalt = sp.simplify(dr2final.subs(1-mu**2, sp.sin(theta)**2))
```
