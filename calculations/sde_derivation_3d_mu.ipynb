**Full derivation of the 3D stochastic differential equations used in `SOLARPROP`.**

Here we derive the SDEs for the three-dimensional case, assuming $\kappa_{r\theta}=\kappa_{\theta{}r}=\kappa_{\theta\phi}=\kappa_{\phi\theta}=0$, and using the variable $\mu=\cos\theta$.

```python
import sympy as sp
```

```python
r, theta, mu, phi = sp.symbols('r theta mu phi', positive=True)
```

```python
vr, vt, vp, vsw = sp.symbols('V_r V_theta V_phi V_sw', positive=True)
```

```python
%config InteractiveShell.ast_node_interactivity='last_expr_or_assign'
```

```python
krr = sp.Function('kappa_rr', positive=True)(r, mu, phi)
```

```python
krp = sp.Function('kappa_r_phi', positive=True)(r, mu, phi)
```

```python
ktt = sp.Function('kappa_theta_theta', positive=True)(r, mu, phi)
```

```python
kpp = sp.Function('kappa_phi_phi', positive=True)(r, mu, phi)
```

We start from Parker's equation:

$$
\frac{\partial f}{\partial t}=\vec{\nabla}\cdot\left(\kappa\cdot\vec{\nabla}f-\vec{V}f\right)+\frac{1}{3p^2}(\vec{\nabla}\cdot\vec{V}_{sw})\frac{\partial p^3f}{\partial p}
$$

where $\vec{V}=\vec{V}_{sw}+\vec{V}_{dr}$.

We write the individual terms in 3D spherical coordinates, and make use of the definitions $\vec{V}_{sw}=V_{sw}\,\vec{e}_r$, $\vec{V}_{dr}\equiv V_{dr}\,\vec{e}_r + V_{d\theta}\,\vec{e}_\theta + V_{d\phi}\,\vec{e}_\phi$ , and $\vec{\nabla}\cdot\vec{V}_{dr}=0$:

$
\begin{align}
\vec{\nabla}\cdot\left(\kappa\cdot\vec{\nabla}f\right)=\frac{1}{r^2}\frac{\partial}{\partial r}\left(r^2\kappa_{rr}\frac{\partial f}{\partial r}+r\kappa_{r\theta}\frac{\partial f}{\partial\theta}+\frac{r\kappa_{r\phi}}{\sin\theta}\frac{\partial f}{\partial\phi}\right)+\frac{1}{r\sin\theta}\frac{\partial}{\partial\theta}\left(\sin\theta\kappa_{\theta r}\frac{\partial f}{\partial r}+\sin\theta\frac{\kappa_{\theta\theta}}{r}\frac{\partial f}{\partial\theta}+\frac{\kappa_{\theta\phi}}{r}\frac{\partial f}{\partial\phi}\right)+\frac{1}{r\sin\theta}\frac{\partial}{\partial\phi}\left(\kappa_{\phi{}r}\frac{\partial f}{\partial{}r} + \frac{\kappa_{\phi\theta}}{r}\frac{\partial f}{\partial\theta} + \frac{\kappa_{\phi\phi}}{r\sin\theta}\frac{\partial f}{\partial\phi}\right),
\end{align}
$

$
\begin{align}
-\vec{\nabla}\cdot\vec{V}f &= -\vec{\nabla}\cdot\vec{V}_{sw}f-\vec{\nabla}\cdot\vec{V}_{dr}f \\
& = -\frac{1}{r^2}\frac{\partial r^2 V_{sw}f}{\partial r}-f( \vec{\nabla}\cdot\vec{V}_{dr} ) - \vec{V}_{dr}\cdot\vec{\nabla}f \\
& =  -\frac{1}{r^2}\frac{\partial(r^2 V_{sw})}{\partial r}f-V_{sw}\frac{\partial f}{\partial r}-V_{dr}\frac{\partial f}{\partial r}-\frac{V_{d\theta}}{r}\frac{\partial f}{\partial\theta}-\frac{V_{d\phi}}{r\sin\theta}\frac{\partial f}{\partial\phi},
\end{align}
$

$
\begin{align}
\frac{1}{3p^2}(\vec{\nabla}\cdot\vec{V}_{sw})\frac{\partial p^3f}{\partial p}&=\frac{1}{r^2}\frac{\partial (r^2V_{sw})}{\partial r}\frac{\partial p^3 f}{\partial p^3}\\
&=\frac{1}{r^2}\frac{\partial (r^2V_{sw})}{\partial r}f + \frac{1}{r^2}\frac{\partial (r^2V_{sw})}{\partial r}p^3\frac{\partial f}{\partial p^3}\\
&=\frac{1}{r^2}\frac{\partial (r^2V_{sw})}{\partial r}f + \frac{p}{3r^2}\frac{\partial (r^2V_{sw})}{\partial r}\frac{\partial f}{\partial p}.
\end{align}
$

Assuming $\kappa_{r\theta}=\kappa_{\theta{}r}=\kappa_{\theta\phi}=\kappa_{\phi\theta}=0$, the first term becomes

$
\begin{align}
\vec{\nabla}\cdot\left(\kappa\cdot\vec{\nabla}f\right)&=\kappa_{rr}\frac{\partial^2 f}{\partial r^2} + \frac{1}{r^2} \frac{\partial (r^2\kappa_{rr})}{\partial r}\frac{\partial f}{\partial r} + 
\frac{\kappa_{r\phi}}{r\sin\theta}\frac{\partial^2f}{\partial{}r\partial\phi}+\frac{1}{r^2\sin\theta}\frac{\partial(r\kappa_{r\phi})}{\partial{}r}\frac{\partial f}{\partial\phi} + 
\frac{\kappa_{\theta\theta}}{r^2}\frac{\partial^2f}{\partial\theta^2}+\frac{1}{r^2\sin\theta}\frac{\partial(\sin\theta\kappa_{\theta\theta})}{\partial\theta}\frac{\partial f}{\partial\theta} +
\frac{1}{r\sin\theta} \frac{\partial\kappa_{\phi r}}{\partial\phi} \frac{\partial f}{\partial r} + 
\frac{\kappa_{\phi r}}{r\sin\theta} \frac{\partial^2 f}{\partial\phi\partial r}+ 
\frac{\kappa_{\phi\phi}}{r^2\sin^2\theta} \frac{\partial^2 f}{\partial\phi^2}+ 
\frac{1}{r^2\sin^2\theta} \frac{\partial\kappa_{\phi\phi}}{\partial\phi} \frac{\partial f}{\partial\phi}.
\\
\end{align}
$



The complete transport equation becomes

$$
\frac{\partial f}{\partial t}=\kappa_{rr}\frac{\partial^2 f}{\partial r^2} + \frac{1}{r^2} \frac{\partial (r^2\kappa_{rr})}{\partial r}\frac{\partial f}{\partial r} + 
\frac{\kappa_{r\phi}}{r\sin\theta}\frac{\partial^2f}{\partial{}r\partial\phi}+\frac{1}{r^2\sin\theta}\frac{\partial(r\kappa_{r\phi})}{\partial{}r}\frac{\partial f}{\partial\phi} + 
\frac{\kappa_{\theta\theta}}{r^2}\frac{\partial^2f}{\partial\theta^2}+\frac{1}{r^2\sin\theta}\frac{\partial(\sin\theta\kappa_{\theta\theta})}{\partial\theta}\frac{\partial f}{\partial\theta} +
\frac{1}{r\sin\theta} \frac{\partial\kappa_{\phi r}}{\partial\phi} \frac{\partial f}{\partial r} + 
\frac{\kappa_{\phi r}}{r\sin\theta} \frac{\partial^2 f}{\partial\phi\partial r}+ 
\frac{\kappa_{\phi\phi}}{r^2\sin^2\theta} \frac{\partial^2 f}{\partial\phi^2}+ 
\frac{1}{r^2\sin^2\theta} \frac{\partial\kappa_{\phi\phi}}{\partial\phi} \frac{\partial f}{\partial\phi}
-V_{sw}\frac{\partial f}{\partial r}-V_{dr}\frac{\partial f}{\partial r}-\frac{V_{d\theta}}{r}\frac{\partial f}{\partial\theta}-\frac{V_{d\phi}}{r\sin\theta}\frac{\partial f}{\partial\phi} +
\frac{p}{3r^2}\frac{\partial (r^2V_{sw})}{\partial r}\frac{\partial f}{\partial p}.
$$


We define 

$$
\mu=\cos\theta,
$$

which leads to

$$
\frac{1}{\sin\theta}\frac{\partial}{\partial\theta}=-\frac{\partial}{\partial\mu}
$$

and thus the transport equation becomes

$$
\frac{\partial f}{\partial t}=\kappa_{rr}\frac{\partial^2 f}{\partial r^2} + \frac{1}{r^2} \frac{\partial (r^2\kappa_{rr})}{\partial r}\frac{\partial f}{\partial r} + 
\frac{\kappa_{r\phi}}{r\sqrt{1-\mu^2}}\frac{\partial^2f}{\partial{}r\partial\phi}+\frac{1}{r^2\sqrt{1-\mu^2}}\frac{\partial(r\kappa_{r\phi})}{\partial{}r}\frac{\partial f}{\partial\phi} + 
\frac{(1-\mu^2)\kappa_{\theta\theta}}{r^2}\frac{\partial^2 f}{\partial\mu^2}+\frac{1}{r^2}\frac{\partial((1-\mu^2)\kappa_{\theta\theta})}{\partial\mu}\frac{\partial f}{\partial\mu} +
\frac{1}{r\sqrt{1-\mu^2}} \frac{\partial\kappa_{\phi r}}{\partial\phi} \frac{\partial f}{\partial r} + 
\frac{\kappa_{\phi r}}{r\sqrt{1-\mu^2}} \frac{\partial^2 f}{\partial\phi\partial r}+ 
\frac{\kappa_{\phi\phi}}{r^2(1-\mu^2)} \frac{\partial^2 f}{\partial\phi^2}+ 
\frac{1}{r^2(1-\mu^2)} \frac{\partial\kappa_{\phi\phi}}{\partial\phi} \frac{\partial f}{\partial\phi}
-V_{sw}\frac{\partial f}{\partial r}-V_{dr}\frac{\partial f}{\partial r}+\frac{\sqrt{1-\mu^2}V_{d\theta}}{r}\frac{\partial f}{\partial\mu}-\frac{V_{d\phi}}{r\sqrt{1-\mu^2}}\frac{\partial f}{\partial\phi} +
\frac{p}{3r^2}\frac{\partial (r^2V_{sw})}{\partial r}\frac{\partial f}{\partial p}.
$$

Therefore, we have expressed the transport equation in the standard form of the backward-in-time Kolmogorov equation:

$$
\frac{\partial f}{\partial t}=\sum\limits_iA_i(\vec{q},t)\frac{\partial f}{\partial q_i}+\frac{1}{2}\sum\limits_{i,j}\left[B(\vec{q},t)B^T(\vec{q},t)\right]_{i,j}\frac{\partial^2 f}{\partial q_i\partial q_j}
$$

with the corresponding SDEs given by

$$
\mathrm{d}\vec{q}=A(\vec{q},t)\mathrm{d}t+B(\vec{q},t)\mathrm{d}W,
$$
where $\mathrm{d}W$ is the increment of the Wiener process.

We use

$$
\vec{q}=(r, \mu, \phi, p)
$$

and from comparing the coefficients, we get

$$
A_r = \frac{1}{r^2}\frac{\partial}{\partial r}(r^2\kappa_{rr}) + \frac{1}{r\sqrt{1-\mu^2}} \frac{\partial\kappa_{r\phi}}{\partial\phi} - V_{sw} - V_{dr}
$$

$$
A_\mu=\frac{1}{r^2}\frac{\partial}{\partial\mu}(\kappa_{\theta\theta}(1-\mu^2))+\frac{\sqrt{1-\mu^2}V_{d\theta}}{r}
$$

$$
A_\phi=\frac{1}{r^2\sqrt{1-\mu^2}}\frac{\partial(r\kappa_{r\phi})}{\partial{}r} + \frac{1}{r^2(1-\mu^2)} \frac{\partial\kappa_{\phi\phi}}{\partial\phi} -\frac{V_{d\phi}}{r\sqrt{1-\mu^2}}
$$

$$
A_p=\frac{p}{3r^2}\frac{\partial r^2V_{sw}}{\partial r}
$$

and

$$
BB^T=
\begin{bmatrix}
2\kappa_{rr} & 0 & 2\frac{\kappa_{r\phi}}{r\sqrt{1-\mu^2}}\\
0 & \frac{2(1-\mu^2)\kappa_{\theta\theta}}{r^2} & 0 \\
2\frac{\kappa_{r\phi}}{r\sqrt{1-\mu^2}}& 0 & 2\frac{\kappa_{\phi\phi}}{r^2(1-\mu^2)}\\
\end{bmatrix}
$$

The elements of the decomposition matrix $B$ are found to be:

```python
B11 = sp.sqrt(2*krr-2*krp**2/kpp)
```

```python
B13 = sp.sqrt(2)*krp/sp.sqrt(kpp)
```

```python
B22 = sp.sqrt(2*(1-mu**2)*ktt)/r
```

```python
B33 = sp.sqrt(2*kpp)/r/sp.sqrt(1-mu**2)
```

```python
B = sp.Matrix([[B11, 0, B13], [0, B22, 0], [0, 0, B33]])
```

```python
print(sp.latex(B))
```

Proof:

```python
sp.simplify(B @ B.T)
```

Dropping the explicit dependence of the diffusion coefficients on $r$, $\mu$, and $\phi$ in the notation, the SDEs are found to be

$$
\mathrm{d}r=\left[\frac{1}{r^2}\frac{\partial}{\partial r}(r^2\kappa_{rr}) + \frac{1}{r\sqrt{1-\mu^2}} \frac{\partial\kappa_{r\phi}}{\partial\phi} - V_{sw} - V_{dr}\right]\,\mathrm{d}t + w_r\, \sqrt{2\kappa_{rr}-\frac{2\kappa_{r\phi}^2}{\kappa_{\phi\phi}}} \,\sqrt{\mathrm{d}t} + w_\phi \frac{\sqrt{2} \kappa_{r \phi}}{\sqrt{\kappa_{\phi\phi}}} \,\sqrt{\mathrm{d}t}
$$

$$
\mathrm{d}\mu = \left[ \frac{1}{r^2}\frac{\partial}{\partial\mu}(\kappa_{\theta\theta}(1-\mu^2))+\frac{\sqrt{1-\mu^2}V_{d\theta}}{r} \right]\,\mathrm{d}t + w_\mu \frac{\sqrt{2(1-\mu^2)\kappa_{\theta\theta}}}{r} \,\sqrt{\mathrm{d}t}
$$

$$
\mathrm{d}\phi = \left[ \frac{1}{r^2\sqrt{1-\mu^2}}\frac{\partial(r\kappa_{r\phi})}{\partial{}r} + \frac{1}{r^2(1-\mu^2)} \frac{\partial\kappa_{\phi\phi}}{\partial\phi} -\frac{V_{d\phi}}{r\sqrt{1-\mu^2}} \right]\,\mathrm{d}t + w_\phi \frac{\sqrt{2\kappa_{\phi\phi}}}{r\sqrt{1-\mu^2}} \,\sqrt{\mathrm{d}t}
$$

$$
\mathrm{d}p=\frac{p}{3r^2}\frac{\partial r^2V_{sw}}{\partial r}\,\mathrm{d}t
$$

where $w_r$, $w_\mu$, and $w_\phi$ are Gaussian random numbers with a mean of zero and unit variance.
