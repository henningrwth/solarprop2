**Calculation of the vector field transformation needed for translating a constant vector -- expressed in polar coordinates -- from one position to another.**

In cartesian coordinates, this is trival: the vector does not change. But in polar coordinates, the $r$-, $\theta$- and $\phi$-components are position-dependent.

```python
import sympy as sp
```

```python
theta, thetaprime, phi, phiprime = sp.symbols('theta thetaprime phi phiprime', real=True)
```

```python
xhat, yhat, zhat = sp.symbols('xhat yhat zhat', real=True)
```

```python
rhat, thetahat, phihat = sp.symbols('rhat thetahat phihat', real=True)
```

```python
Ax, Ay, Az, Ar, Atheta, Aphi, Arprime, Athetaprime, Aphiprime = sp.symbols('A_x A_y A_z A_r A_theta A_phi Aprime_r Aprime_theta Aprime_phi', real=True)
```

```python
%config InteractiveShell.ast_node_interactivity='last_expr_or_assign'
```

```python
M = sp.Matrix([[sp.sin(theta)*sp.cos(phi), sp.sin(theta)*sp.sin(phi), sp.cos(theta)], [sp.cos(theta)*sp.cos(phi), sp.cos(theta)*sp.sin(phi), -sp.sin(theta)], [-sp.sin(phi), sp.cos(phi), 0]])
```

```python
n = sp.Matrix([xhat, yhat, zhat])
```

```python
spherical_basis = M@n
```

```python
Ar*rhat + Atheta*thetahat + Aphi*phihat
```

```python
Aprime = sp.expand(Ar*spherical_basis[0] + Atheta*spherical_basis[1] + Aphi*spherical_basis[2])
```

```python
Ax*xhat + Ay*yhat + Az*zhat
```

```python
n_spherical = sp.Matrix([rhat, thetahat, phihat])
```

```python
cartesian_basis = M.T@n_spherical
```

```python
A_spherical = sp.expand(Ax*cartesian_basis[0] + Ay*cartesian_basis[1] + Az*cartesian_basis[2]) 
```

```python
Fr = A_spherical.coeff(rhat, 1)
```

```python
Ftheta = A_spherical.coeff(thetahat, 1)
```

```python
Fphi = A_spherical.coeff(phihat, 1)
```

```python
sp.simplify(Fr*spherical_basis[0] + Ftheta*spherical_basis[1] + Fphi*spherical_basis[2])
```

```python
prime_subs = [(Ar, Arprime), (Atheta, Athetaprime), (Aphi, Aphiprime), (theta, thetaprime), (phi, phiprime)]
```

```python
Ax_prime = Aprime.coeff(xhat, 1).subs(prime_subs)
```

```python
Ay_prime = Aprime.coeff(yhat, 1).subs(prime_subs)
```

```python
Az_prime = Aprime.coeff(zhat, 1).subs(prime_subs)
```

```python
A_point = sp.simplify(A_spherical.subs([(Ax, Ax_prime), (Ay, Ay_prime), (Az, Az_prime)]))
```

```python
A_point_r = A_point.coeff(rhat, 1)
```

```python
A_point_theta = A_point.coeff(thetahat, 1)
```

```python
A_point_phi = A_point.coeff(phihat, 1)
```
